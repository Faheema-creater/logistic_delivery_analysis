##**2. Dataset Collection and Description**

###*Dataset Source*

The dataset used in this project is the Brazilian E-Commerce Public Dataset by Olist, a publicly available e-commerce dataset containing information about orders, customers, products, payments, freight, reviews, and delivery performance.
For this preprocessing task, the Orders dataset is used because it contains important logistics-related information such as order status, purchase timestamp, approval timestamp, actual delivery timestamp, and estimated delivery date.

Data quality is assessed by checking missing values, duplicate records, categorical distributions, and basic statistical information. These checks help identify problems that need to be addressed before performing further analysis.

##**7. Data Type Conversion**

The timestamp columns are initially stored as text values. For logistics analysis, these columns must be converted into datetime format so that date differences and delivery durations can be calculated correctly.
The errors="coerce" option converts invalid date values into missing values instead of causing the program to stop. This allows invalid or incorrectly formatted dates to be identified during the data-quality assessment.

```python
date_columns = [
    "order_purchase_timestamp",
    "order_approved_at",
    "order_delivered_timestamp",
    "order_estimated_delivery_date"
]

for column in date_columns:
    orders[column] = pd.to_datetime(
        orders[column],
        errors="coerce"
    )

print("Data types after conversion:")
orders[date_columns].dtypes
orders.info()
```

##**8. Missing Value Analysis**

Missing values are common in real-world logistics datasets. They must be analyzed carefully because replacing or removing missing values without understanding their meaning can introduce bias.
In this dataset, missing values may occur in approval or delivery timestamps. A missing delivery timestamp can be meaningful because an order may not have reached the delivered stage.
Therefore, missing values are first identified and analyzed rather than automatically replaced.

###*Treatment Strategy*

Missing approval timestamps will be retained because only a small number of records are affected.

Missing delivery timestamps will be retained when they correspond to orders that have not been delivered.

Missing values will not be replaced with arbitrary values because doing so could create false delivery information.

```python
print("Missing values after date conversion:")
print(orders.isnull().sum())
```

```python
print("Missing delivery timestamps by order status:")
print(
    orders[orders["order_delivered_timestamp"].isnull()]
    ["order_status"]
    .value_counts()
)
```

##**9. Duplicate Record Detection**

Duplicate records can lead to double-counting and incorrect logistics performance measurements. Therefore, duplicate rows are checked before further analysis.
If duplicate records are found, they can be removed using pandas drop_duplicates() method. In this dataset, the duplicate count is examined before deciding whether removal is necessary.

```python
duplicate_count = orders.duplicated().sum()

print("Number of duplicate rows:", duplicate_count)
```

##**10. Data Consistency Validation**

Data consistency checks are performed to identify values that may be logically incorrect for logistics analysis.
The following conditions are examined:

Negative delivery durations

Extremely long delivery durations

Delivered orders without delivery timestamps

Unexpected relationships between order status and delivery information

These checks help distinguish genuine operational variations from potentially invalid records.

```python
delivered_without_timestamp = (
    (orders["order_status"] == "delivered") &
    (orders["order_delivered_timestamp"].isnull())
).sum()

print(
    "Delivered orders without delivery timestamp:",
    delivered_without_timestamp
)
```

```python
orders["delivery_days"] = (
    orders["order_delivered_timestamp"]
    - orders["order_purchase_timestamp"]
).dt.days

print("Negative delivery durations:")
print((orders["delivery_days"] < 0).sum())

print("\nDelivery durations greater than 100 days:")
print((orders["delivery_days"] > 100).sum())
```

##**11. Feature Engineering**

A new feature named delivery_days is created to represent the number of days between the order purchase date and the actual delivery date.
This feature is useful for logistics analysis because it converts timestamp information into a numerical measure that can be analyzed statistically.
The calculation is performed only when both purchase and delivery timestamps are available.

```python
orders[[
    "order_purchase_timestamp",
    "order_delivered_timestamp",
    "delivery_days"
]].head(10)
```

```python
print("Delivery time statistics:")
display(orders["delivery_days"].describe())
```

##**12. Outlier Detection**

Outliers are observations that are unusually different from the majority of the data. In logistics, extremely long delivery times may indicate operational delays, unusual circumstances, or data-quality problems.
The Interquartile Range (IQR) method is used to identify potential outliers.
The IQR is calculated as:
IQR = Q3 − Q1
The lower and upper boundaries are calculated as:
Lower Bound = Q1 − 1.5 × IQR
Upper Bound = Q3 + 1.5 × IQR
Records outside these boundaries are considered potential outliers.

```python
Q1 = orders["delivery_days"].quantile(0.25)
Q3 = orders["delivery_days"].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)
```

```python
outliers = orders[
    (orders["delivery_days"] < lower_bound) |
    (orders["delivery_days"] > upper_bound)
]

print("Number of potential outliers:", len(outliers))
```

##**13. Outlier Treatment Strategy**

Potential outliers are not automatically removed because extreme delivery times can represent genuine logistics problems.
For example, an unusually long delivery duration may indicate a delayed shipment, transportation issue, operational bottleneck, or exceptional order.
Therefore, the identified outliers are retained for the current logistics analysis unless a record is proven to be invalid. Instead of deleting potentially valuable information, the outlier records can be flagged for further investigation.
This approach prevents genuine operational problems from being hidden during analysis.

```python
orders["delivery_outlier"] = (
    (orders["delivery_days"] < lower_bound) |
    (orders["delivery_days"] > upper_bound)
)

print("Outlier flag distribution:")
print(orders["delivery_outlier"].value_counts())
```

##**14. Data Normalization**

Normalization is used to transform numerical values to a common scale. Min-Max normalization converts values into a range between 0 and 1.
The formula is:
Normalized Value = (X − Minimum) / (Maximum − Minimum)
Normalization can be useful when numerical variables with different scales are later used in machine-learning algorithms or other analytical techniques.
The normalized delivery duration is stored in a separate column so that the original delivery-time information is preserved.

```python
delivery_min = orders["delivery_days"].min()
delivery_max = orders["delivery_days"].max()

orders["delivery_days_normalized"] = (
    (orders["delivery_days"] - delivery_min) /
    (delivery_max - delivery_min)
)

orders[
    ["delivery_days", "delivery_days_normalized"]
].head(10)
```

```python
print(
    "Minimum normalized value:",
    orders["delivery_days_normalized"].min()
)

print(
    "Maximum normalized value:",
    orders["delivery_days_normalized"].max()
)
```
